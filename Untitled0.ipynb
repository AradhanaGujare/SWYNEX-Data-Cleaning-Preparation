```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")
df.to_csv("original_titanic.csv", index=False)
print(df.head())
```

```python
print("Rows and columns:", df.shape)
print()
print("Missing values in each column:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
print()
print("Data types:")
print(df.dtypes)
```
